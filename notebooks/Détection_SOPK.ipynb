## Modélisation 

### Division de données en : Données d'entrée (Input) et Données de sorties (Output)

```python
import numpy as np 
x= data.drop('PCOS (Y/N)' , axis=1) 
x_list = list(x.columns)
x =np.array(x)
y= np.array(data['PCOS (Y/N)'])

print(x_list)
print(x)
print(y)
```

```python
x_list
```

### 2) Division de données en : Données d'Apprentissage (Data Training : 90%) et Données de Test (Data Testing : 10%)

```python
from sklearn.model_selection import train_test_split 
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.1 , random_state=40)
# Division du Test en Test (15%) et Final Test (15%)
x_test, x_final, y_test, y_final = train_test_split(x_test, y_test, test_size=0.5, random_state=42)


print('Training Data x : ' , x_train.shape)
print('Training Data y: ' , y_train.shape)
print('Testing Data x' , x_test.shape)
print('Testing Data y : ' , y_test.shape)

```

```python
# Sélectionner les 5 dernières lignes pour le Final Test
x_final_test = x_test[-5:]  
y_final_test = y_test[-5:]

# Mettre à jour le Test Set après suppression des 5 dernières lignes
x_test = x_test[:-5]
y_test = y_test[:-5]

# Vérification des tailles après extraction
print("Testing Data:", x_test.shape, y_test.shape)
print("Final Testing Data:", x_final_test.shape, y_final_test.shape)
```

```python
#Optimisation
from collections import Counter
from imblearn.over_sampling import SMOTE

print("Répartition des classes avant SMOTE:", Counter(y))

smote = SMOTE(sampling_strategy='auto', random_state=42)  # auto = équilibrer les classes
x_resampled, y_resampled = smote.fit_resample(x_train, y_train)

print("Répartition des classes après SMOTE:", Counter(y_resampled))
```

## 5) Comparer les modéles : Gradient Boosting - XGBoost - Random Forest

```python
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import recall_score
import xgboost as xgb

# Créer les modèles
models = {
    "Random Forest": RandomForestClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(),
    "XGBoost": xgb.XGBClassifier()
}
```

```python

# Calculer la précision pour chaque modèle
accuracies = {}


for name, model in models.items():
    model.fit(x_resampled, y_resampled)
    y_pred = model.predict(x_test)

    accuracy = accuracy_score(y_test, y_pred)
    accuracies[name] = accuracy


# Afficher la comparaison sous forme de barre
plt.figure(figsize=(10, 6))
plt.bar(accuracies.keys(), accuracies.values(), color=['blue', 'green', 'red'])
plt.title('Précision des Modèles')
plt.xlabel('Modèle')
plt.ylabel('Précision')
plt.yticks(list(accuracies.values()))
plt.ylim(0, 1)  # Limiter l'axe y entre 0 et 1
plt.show()
```

```python
# Calculer le rappel pour chaque modèle  
# Il représente la capacité d’un modèle de classification à identifier correctement toutes les cas positives
# Vrai Positif (VP) : Le patient est malade et le test dit qu’il est malade 
# Vrai Négatif (VN) : Le patient est sain et le test dit qu’il est sain 
# Un rappel élevé signifie que le modèle détecte la majorité des cas positifs, même s’il génère plus de faux positifs
# Faux Positif (FP) : Le patient est sain mais le test dit qu’il est malade 


recalles = {}

for name, model in models.items():
    model.fit(x_resampled, y_resampled)
    y_pred = model.predict(x_test)
    recall = recall_score(y_test, y_pred)
    recalles [name] = recall


# Afficher la comparaison sous forme de barre
plt.figure(figsize=(10, 6))
plt.bar(recalles.keys(), recalles.values(), color=['blue', 'green', 'red'])
plt.title('Rappel des Modèles')
plt.xlabel('Modèle')
plt.ylabel('Rappel')
plt.yticks(list(recalles.values()))
plt.ylim(0, 1)  # Limiter l'axe y entre 0 et 1
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
import xgboost as xgb
# Créer les modèles
models = {
    "Random Forest": RandomForestClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(),
    "XGBoost": xgb.XGBClassifier()
}

# Tracer les matrices de confusion
fig, axs = plt.subplots(1, 3, figsize=(18, 6))

for i, (name, model) in enumerate(models.items()):
    model.fit(x_resampled, y_resampled)
    y_pred = model.predict(x_test)
    
    # Calculer la matrice de confusion
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Classe 0", "Classe 1"])
    
    # Afficher la matrice de confusion pour chaque modèle
    disp.plot(ax=axs[i], cmap="Blues")
    axs[i].set_title(f'Matrice de Confusion - {name}')

# Afficher toutes les matrices
plt.tight_layout()
plt.show()
```
